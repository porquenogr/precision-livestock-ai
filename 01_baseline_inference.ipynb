```python
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow
model = YOLO('yolov8n.pt')
results = model('https://ichef.bbci.co.uk/ace/standard/1800/cpsprodpb/4fc5/live/2a79b280-372d-11f0-aecb-17d44f25a860.png')
for r in results:
    im_bgr = r.plot()
    cv2_imshow(im_bgr)
```
