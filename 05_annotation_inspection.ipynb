```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "trainingdatapro/farm-animals-pigs-detection-dataset"
)

print("[INFO] Dataset downloaded to:", dataset_path)
```

```python
from pathlib import Path

source_path = Path(dataset_path)

print("[INFO] Dataset exists:", source_path.exists())

for item in source_path.iterdir():
    print(" -", item.name)
```

```python
xml_file = source_path / "annotations.xml"

print("[INFO] XML exists:", xml_file.exists())
```

```python
with open(xml_file, "r", encoding="utf-8") as f:
    for _ in range(20):
        print(f.readline().strip())
```

```python
from xml.etree import ElementTree as ET

tree = ET.parse(xml_file)
root = tree.getroot()

labels = root.find("meta/task/labels")

for label in labels:
    print(label.find("name").text)
```

```python
images = root.findall("image")

print("[INFO] Number of annotated images:", len(images))

first_image = images[0]

print("\n[INFO] First image:")
print("Name:", first_image.get("name"))
print("Width:", first_image.get("width"))
print("Height:", first_image.get("height"))

print("\n[INFO] Objects:")
for box in first_image.findall("box"):
    print("Label:", box.get("label"))
    print("xtl:", box.get("xtl"))
    print("ytl:", box.get("ytl"))
    print("xbr:", box.get("xbr"))
    print("ybr:", box.get("ybr"))
```
